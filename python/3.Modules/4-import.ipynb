### Importing Modules in Python: Modules and Packages

In python modules and packages help organize and reuse code. Here's comprehensive guide on how to import them.

```python
import math
math.sqrt(16)
```

```python
from math import sqrt,pi
print(sqrt(25))
print(pi)
```

```python
import numpy as np ## np is the aliars here
np.array([1,2,3,5])
```

```python
from math import * ## all the modules in the packages are called at once
print(sqrt(25))
print(pi)
```

```python
from package.maths import addition
addition(2,3) ## One thing to remember is that we have to add the package file in the same folder then only it can be executed.
```

```python
from package import maths
maths.addition(2,3)
```
