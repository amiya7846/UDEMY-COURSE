#### Definition
- A function is a block of code that performs a specific task. Fuction help in organising code, reusing code, and improve readability.

```python
## syntax
def function_name(parameters):
    """Docstring""" ## It describes what the function does
    # Function body
    return expression  
```

```python
## Why use function?
num = 24
if num % 2 == 0:
    print(f"{num} is even")
else:
    print(f"{num} is odd")    
```

```python
def even_or_odd(num):
    """This function checks if a number is even or odd"""
    if num % 2 == 0:
        print(f"{num} is even")
    else:
        print(f"{num} is odd")
        
```

```python
## call this function
even_or_odd(24)
```

```python
## Functions with multiple parameters
def add_numbers(num1, num2):
    """This function adds two numbers"""
    num3 = num1 + num2
    return num3
result = add_numbers(10,20)
print(result)
```

```python
def greet(name):
    print(f"hello {name} Welcome to the paradise")

greet("Amiya") ## it will take the value passed to the function
```

```python
## Default parameters
def greet(name = "Guest"):
    print(f"hello {name} Welcome to the paradise")

greet() ## it will take default value    
```

```python
## Variable length arguments
## positional arguments 

def print_numbers(*args): ## Itneednot be named args, it can be any name like amiya or something else
    for num in args:
        print(num)
```

```python
print_numbers(1,2,3,4,5,6,7,"Amiya") ## It can take any number of arguments
```

```python
## Keyword arguments
def print_details(**kwargs): ## Itneednot be named args, it can be any name like amiya or something else
    for key, value in kwargs.items():
        print(f"{key} : {value}")
```

```python
print_details(name = "Amiya", age = 20, city = "Bengaluru")
```

```python
def print_info(*args, **kwargs):
    for val in args:
        print(f" Positional argument :{val}")

    for key, value in kwargs.items():
        print(f"{key}:{value}")
```

```python
print_info(1,2,3, name = "Amiya", age = 20, city = "Bengaluru")
```

```python
## Return multiple parameters from a function 
def multiply(num1,num2):
    """This function multiplies two numbers"""
    return num1 * num2, num1 
multiply(10,20)
```
