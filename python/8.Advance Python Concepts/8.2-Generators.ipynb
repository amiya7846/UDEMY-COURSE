### Generators

Generators are a simpler way to create iterators. They use the yeild keyword to produce a series of values lazily,which means they generate values on the fly and donot store them in memory.

```python
def square(n):
    for i in range(3):
        yield i**2  ## The yield statement is used to produce a series of values over time, allowing the function to return a generator object that can be iterated over. In this case, it will yield the squares of the numbers 0, 1, and 2.
```

```python
square(3)
```

```python
for i in square(3):
    print(i)
```

```python
a = square(3)
a
```

```python
next(a) ## first it will print 0 then 1 and then 4 and after that it will show the error message.
```

```python
def my_generator():
    yield 1
    yield 2
    yield 3
```

```python
gen=my_generator()
gen
```

```python
next(gen)
```

```python
for val in gen:
    print(val)
```

### Practical Example:Reading the large files

Generators are partially useful for reading the large files because they allow you to process one line at a time without loading the entire file into memory.

```python
### Practical : Reading Large files 

def read_large_file(file_path):
    with open(file_path,'r') as file:
        for line in file:
            yield line
```

```python
file_path = 'large_file.txt'
for line in read_large_file(file_path):
    print(line.strip())
```
