### Decorators 

Decorators are a powerful and flexible feature in python that allows you to modify the behavior of a function or class method. They  are commonly used to add functionality to functions or methods without modifying their actual code. 

```python
## Function copy 
## Closures 
## Decorators 
```

```python
## Function copy 
def welcome():
    return "Welcome to the advance python course"
welcome()
```

```python
wel = welcome ## Copy the welcome function to the wel
print(wel())  ## Then print the wel function 
del welcome   ## then delete the welcome function
print(wel())  ## again print the wel function 
```

```python
##closure means a function inside another function 
def main_welcome():
    msg = "welcome"
    def sub_welcome_method():
        print("Welcome to the advanced python course")
        print(msg) ## Here we can also access the main method in the sub-method 
        print("Please learn the concepts properly")
    return sub_welcome_method()
```

```python
main_welcome()
```

```python
##closure means a function inside another function 
def main_welcome(func):
    msg = "welcome"
    def sub_welcome_method():
        print("Welcome to the advanced python course")
        func("Welcome everyone to this tutorial") 
        print("Please learn the concepts properly")
    return sub_welcome_method()
```

```python
main_welcome(print)
```

```python
##closure means a function inside another function 
def main_welcome(func,lst):
    msg = "welcome"
    def sub_welcome_method():
        print("Welcome to the advanced python course")
        print(func(lst))
        print("Please learn the concepts properly")
    return sub_welcome_method()
```

```python
main_welcome(len,[1,2,3,4,5,6,7])
```

```python
### Decorators 
##closure means a function inside another function 
def main_welcome(func):
    def sub_welcome_method():
        print("Welcome to the advanced python course")
        func() 
        print("Please learn the concepts properly")
    return sub_welcome_method()
```

```python
def course_introduction():
    print("This is an advanced python course")
course_introduction()
```

```python
main_welcome(course_introduction)
```

```python
@main_welcome
def course_introduction():
    print("This is an advanced python course")
```

```python
## Decorator
def my_decorator(func):
    def wrapper():
        print("something is happening before the function is called.")
        func()
        print("somrthing is happening after the function is called.")
    return wrapper
```

```python
@my_decorator
def say_hello():
    print("Hello!")
```

```python
say_hello()
```

```python
## Decorators with arguments 
def repeat(n):
    def decorator(func):
        def wrapper(*args, **kwargs):
            for _ in range(n):
                func(*args, **kwargs)
        return wrapper
    return decorator
```

```python
@repeat(5)
def say_hello():
    print("Hello!")
```

```python
say_hello()
```
