### Iterators 

Iterators are the advanced Python concepts that allow for efficient looping and memory management.Iterators provide a way to access elements of a collection sequentially without exposing the underlying structure.

```python
my_list = [1,2,3,4,5,6]
for i in my_list:
    print(i)
```

```python
type(my_list)
```

```python
print(my_list)
```

```python
## Iterator
Iterator = iter(my_list)
print(type(Iterator))
```

```python
Iterator
```

```python
## In order to iterate through all the elements we need to use the Next keyword
## and it will show the first item in the list that is in our case is 1
next(Iterator)
```

```python
## And if we execute it again it will go to the next elements that is 2 and it will continue 
next(Iterator)
```

```python
## And after coming to the last element it will show an error that stopIteration
next(Iterator)
```

```python
Iterator = iter(my_list)
try:
    print(next(Iterator))
except StopIteration:
    print("There are no elements in the iterator")
```

```python
## String Iterator 
my_string = "Hello"
string_iterator = iter(my_string)
print(next(string_iterator)) #Output:H
print(next(string_iterator)) #Output:e
```
